```python
import pandas as pd
sonar_data= pd.read_csv('/content/Copy of sonar data.csv',header=None)
```

```python
print(sonar_data.shape)
print(sonar_data.describe())
```

```python
print(sonar_data[60].value_counts())
```

```python
x=sonar_data.iloc[:, 0:60]
y=sonar_data.iloc[:, 60]
```

```python
from sklearn.model_selection import train_test_split, cross_val_score
x_train,x_test,y_train,y_test=train_test_split(x,y,stratify=y,test_size=0.1,random_state=1)

```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
#x_train_scaled=scaler.fit_transform(x_train,y_train)
#x_test_scaled=scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression(C=0.01)
#model.fit(x_train_scaled,y_train)
#y_predict=model.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score, precision_score
#print(accuracy_score(y_test,y_predict))
```

```python
#print(recall_score(y_test,y_predict,pos_label='M'))
```

```python
#print(precision_score(y_test,y_predict,pos_label='M'))
```

```python
from sklearn.pipeline import Pipeline
pipeline= Pipeline([('scaler',scaler),('model',model)])
```

```python
cv_scores=cross_val_score(pipeline, x_train, y_train , cv=5)
```

```python
print(cv_scores)
```

```python
import numpy as np
print(np.mean(cv_scores))
```

```python
pipeline.fit(x_train,y_train)
```

```python
y_train_predict=pipeline.predict(x_train)
```

```python
print(accuracy_score(y_train,y_train_predict))
```

```python
y_predict=pipeline.predict(x_test)
```

```python
print(accuracy_score(y_test,y_predict))
```

```python
print(confusion_matrix(y_test,y_predict,labels=['M','R']))
```

```python
print(recall_score(y_test,y_predict,pos_label='M'))
print(precision_score(y_test,y_predict,pos_label='M'))
```
